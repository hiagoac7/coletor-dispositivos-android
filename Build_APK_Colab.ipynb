```
import json, getpass
from pathlib import Path
url = getpass.getpass('URL HTTPS do Apps Script: ')
nomes = getpass.getpass('Usuarios separados por ponto e virgula: ')
Path('config.json').write_text(json.dumps({'sheets_url': url, 'usuarios': [n.strip() for n in nomes.split(';') if n.strip()]}, ensure_ascii=False), encoding='utf-8')
```

```
# Execute depois de preparar Buildozer, Java, SDK/NDK e dependencias do sistema.
!buildozer android debug
```
